## Introducción

El **Razonamiento Basado en Casos (CBR)** es un paradigma de IA que resuelve nuevos problemas reutilizando soluciones de problemas similares del pasado. El ciclo CBR se compone de cuatro etapas:

1. **Retrieve** – Recuperar los casos más similares al problema actual.
2. **Reuse** – Adaptar la solución del caso recuperado al nuevo problema.
3. **Revise** – Evaluar y corregir la solución propuesta.
4. **Retain** – Almacenar el nuevo caso resuelto en la base de casos.

En esta práctica trabajaremos con la librería **CBRkit** para construir un sistema CBR aplicado a la **recomendación de coches de segunda mano**. Dado el perfil de un usuario que busca un coche, el sistema recuperará los vehículos más similares de la base de casos y adaptará el precio estimado.

## Parte 3: Métricas de similitud

CBRkit ofrece métricas de similitud para distintos tipos de datos en el módulo `cbrkit.sim`. La **similitud global** entre dos casos se calcula combinando las similitudes locales de cada atributo mediante una función de agregación (normalmente la media ponderada).

### 3.1 Métricas para atributos numéricos (`cbrkit.sim.numbers`)

Para atributos como `year` y `odometer` usaremos métricas numéricas.

```python
# Exploración de las métricas numéricas disponibles
# cbrkit.sim.numbers.linear: similitud lineal normalizada por un rango máximo
# cbrkit.sim.numbers.threshold: 1.0 si la diferencia es menor que un umbral, 0.0 en caso contrario

# Métrica lineal: sim = max(0, 1 - |x - y| / max_distance)
year_sim_linear = cbrkit.sim.numbers.linear(max=10)  # rango máximo de 10 años
odometer_sim_linear = cbrkit.sim.numbers.linear(max=150000)  # rango máximo de 150,000 km

# Prueba manual de las métricas
print("=== Similitud de AÑO (lineal, max=10) ===")
pairs_year = [(2018, 2018), (2018, 2019), (2018, 2020), (2018, 2015), (2018, 2010)]
for x, y in pairs_year:
    sim = year_sim_linear(x, y)
    print(f"  sim({x}, {y}) = {sim:.3f}")

print("\n=== Similitud de KILOMETRAJE (lineal, max=150000) ===")
pairs_odo = [(50000, 50000), (50000, 60000), (50000, 80000), (50000, 120000), (50000, 200000)]
for x, y in pairs_odo:
    sim = odometer_sim_linear(x, y)
    print(f"  sim({x:,}, {y:,}) = {sim:.3f}")
```

```python
def plot_year(series, labels, years, reference_year):
    plt.figure(figsize=(10, 5))
    for s in series:
        plt.plot(years, s, marker='o', label=labels[series.index(s)])
    plt.xticks(ticks=years, labels=years)
    plt.title(f"Similitud de AÑO respecto a {reference_year}")
    plt.xlabel("Año comparado")
    plt.ylabel("Similitud")
    plt.ylim(-0.05, 1.05)
    plt.legend()
    plt.grid()
    plt.show()
```

```python
reference_year = 2018
years = list(range(2008, 2024))
year_sims = [year_sim_linear(reference_year, y) for y in years]
plot_year([year_sims], ['Lineal'], years, reference_year)
```

### 3.2 Métricas para atributos categóricos (`cbrkit.sim.generic`)

Para atributos como `fuel`, `transmission` o `condition` podemos usar:
- **`equality`**: 1.0 si son iguales, 0.0 si no lo son.
- **`table`**: define similitudes personalizadas entre pares de valores.
- Una función personalizada.

```python
# Métrica de igualdad (la más básica para categóricos)
equality_sim = cbrkit.sim.generic.equality()

print("=== Similitud de COMBUSTIBLE (igualdad) ===")
fuels = [("gas", "gas"), ("gas", "electric"), ("gas", "hybrid"), ("electric", "electric")]
for x, y in fuels:
    print(f"  sim({x}, {y}) = {equality_sim(x, y):.1f}")
```

```python
# Métrica de tabla: permite definir similitudes parciales entre valores categóricos
# Útil cuando hay una relación semántica entre las categorías

# Para el atributo 'condition', hay un orden natural: excellent > good > fair > salvage
condition_sim_table = cbrkit.sim.generic.table(
    {
        ("excellent", "excellent"): 1.0,
        ("excellent", "good"):      0.7,
        ("excellent", "fair"):      0.3,
        ("excellent", "salvage"):   0.0,
        ("good",      "excellent"): 0.7,
        ("good",      "good"):      1.0,
        ("good",      "fair"):      0.5,
        ("good",      "salvage"):   0.1,
        ("fair",      "excellent"): 0.3,
        ("fair",      "good"):      0.5,
        ("fair",      "fair"):      1.0,
        ("fair",      "salvage"):   0.2,
        ("salvage",   "excellent"): 0.0,
        ("salvage",   "good"):      0.1,
        ("salvage",   "fair"):      0.2,
        ("salvage",   "salvage"):   1.0,
    },
    default=0.0
)

print("=== Similitud de CONDICIÓN (tabla personalizada) ===")
conditions = ["excellent", "good", "fair", "salvage"]
for c1 in conditions:
    for c2 in conditions:
        sim = condition_sim_table(c1, c2)
        print(f"  sim({c1:10s}, {c2:10s}) = {sim:.1f}")
```

```python
# Visualizar la matriz de similitud para 'condition'
conditions = ["excellent", "good", "fair", "salvage"]
matrix = [[condition_sim_table(c1, c2) for c2 in conditions] for c1 in conditions]
matrix_df = pd.DataFrame(matrix, index=conditions, columns=conditions)

plt.figure(figsize=(6, 5))
sns.heatmap(matrix_df, annot=True, fmt=".1f", cmap="YlOrRd", vmin=0, vmax=1)
plt.title("Matriz de similitud para 'condition'")
plt.tight_layout()
plt.show()
```

### **Ejercicio 2:** Métrica para el atributo `fuel`

La igualdad exacta para `fuel` puede ser demasiado estricta. Por ejemplo, `hybrid` es más cercano a `electric` que a `gas`.

1. Crea una **métrica de tabla** para el atributo `fuel` que tenga en cuenta estas similitudes parciales.
2. Visualiza la matriz de similitud resultante.
3. Justifica tus elecciones de valores en una celda de texto.

```python
# Ejercicio 2: Tu código aquí
# Valores posibles de fuel: gas, electric, hybrid, diesel, other

fuel_sim = cbrkit.sim.generic.table(
    {
        ("hybrid", "electric"): 0.7,
        ("electric", "hybrid"): 0.7,
        ("hybrid", "gas"): 0.6,
        ("gas", "hybrid"): 0.6,
        ("hybrid", "diesel"): 0.5,
        ("diesel", "hybrid"): 0.5,
        ("gas", "diesel"): 0.6,
        ("diesel", "gas"): 0.6,
        ("gas", "electric"): 0.1,
        ("electric", "gas"): 0.1,
        ("diesel", "electric"): 0.1,
        ("electric", "diesel"): 0.1,
        ("other", "gas"): 0.2,
        ("gas", "other"): 0.2,
        ("other", "electric"): 0.2,
        ("electric", "other"): 0.2,
        ("other", "hybrid"): 0.2,
        ("hybrid", "other"): 0.2,
        ("other", "diesel"): 0.2,
        ("diesel", "other"): 0.2,
    },
    default=0.0
)

import matplotlib.pyplot as plt
import numpy as np

fuel_types = ["gas", "diesel", "hybrid", "electric", "other"]
matrix = np.zeros((len(fuel_types), len(fuel_types)))

for i, f1 in enumerate(fuel_types):
    for j, f2 in enumerate(fuel_types):
        matrix[i][j] = fuel_sim(f1, f2)

fig, ax = plt.subplots(figsize=(6, 5))
im = ax.imshow(matrix, cmap="YlOrRd", vmin=0, vmax=1)
ax.set_xticks(range(len(fuel_types)))
ax.set_yticks(range(len(fuel_types)))
ax.set_xticklabels(fuel_types)
ax.set_yticklabels(fuel_types)

for i in range(len(fuel_types)):
    for j in range(len(fuel_types)):
        ax.text(j, i, f"{matrix[i][j]:.1f}", ha="center", va="center", color="black")

plt.colorbar(im)
plt.title("Matriz de similitud - Fuel")
plt.tight_layout()
plt.show()
```

**Justificación del Ejercicio 2:**

Los valores de similitud se han elegido en función de la proximidad tecnológica entre los tipos de combustible:

- **hybrid / electric = 0.7**: Los híbridos comparten motor eléctrico con los eléctricos puros, por lo que un comprador interesado en un eléctrico podría considerar un híbrido como alternativa viable.
- **hybrid / gas = 0.6**: Los híbridos también tienen motor de combustión, así que guardan relación con los de gasolina, aunque menor que con los eléctricos.
- **gas / diesel = 0.6**: Ambos son combustibles fósiles con tecnología de motor de combustión interna similar, pero el comportamiento de conducción y costes de mantenimiento difieren.
- **gas / electric y diesel / electric = 0.1**: Son tecnologías muy distintas (combustión vs. eléctrico), con costes de mantenimiento, rendimiento y experiencia de conducción diferentes.
- **other = 0.2 con todos**: El tipo "other" es ambiguo, por lo que se asigna una similitud baja pero no nula con todos los demás.
- **default = 0.0**: Si aparece algún par no contemplado, se asume similitud nula.

```python
# Métricas de similitud de cadenas disponibles en CBRkit
# Necesitamos instalar la dependencia 'nlp' de cbrkit para Levenshtein y Jaro
%pip install levenshtein --quiet
```

```python

# Métrica de Levenshtein: basada en distancia de edición
levenshtein_sim = cbrkit.sim.strings.levenshtein()

# Métrica de Jaro-Winkler: mejor para nombres y cadenas cortas
jaro_sim = cbrkit.sim.strings.jaro_winkler()

print("=== Comparación de métricas de STRING para 'model' ===")
model_pairs = [
    ("camry se", "camry se"),       # igual
    ("camry se", "camry le"),       # muy similar (mismo modelo, trim diferente)
    ("camry se", "camry xse"),      # similar
    ("camry se", "corolla le"),     # misma marca, modelo diferente
    ("camry se", "f-150 xlt"),      # completamente diferente
]

print(f"{'Par':<35} {'Levenshtein':>12} {'Jaro-Winkler':>12}")
print("-" * 62)
for x, y in model_pairs:
    lev = levenshtein_sim(x, y)
    jaro = jaro_sim(x, y)
    print(f"  ({x!r}, {y!r})   {lev:>12.3f} {jaro:>12.3f}")
```

```python
# También podemos definir una función personalizada
# CBRkit exige que los parámetros se llamen x e y

def model_sim_prefix(x: str, y: str) -> float:
    """
    Similitud para el modelo del coche:
    - Coincidencia exacta: 1.0
    - La primera palabra (nombre de modelo) coincide: 0.7
    - La primera palabra NO coincide: basado en Jaro
    """
    if x == y:
        return 1.0
    x_parts = x.lower().split()
    y_parts = y.lower().split()
    if x_parts[0] == y_parts[0]:
        return 0.7
    # Si las primeras palabras son distintas, similitud baja
    return 0.1

print("=== Función personalizada 'model_sim_prefix' ===")
for x, y in model_pairs:
    sim = model_sim_prefix(x, y)
    print(f"  ({x!r}, {y!r}) => {sim:.2f}")
```

```python
# Ejercicio 3: Comparación de retrievers sobre las tres consultas

def run_retriever(casebase, query, retriever, label):
    result = cbrkit.retrieval.apply.apply_query(casebase, query, retriever)
    step = result.steps[-1]
    qstep = list(step.queries.values())[0]
    rows = []
    for case_id in list(qstep.ranking)[:3]:
        sim_val = float(qstep.similarities[case_id].value)
        case = casebase[case_id]
        rows.append({
            "Retriever": label,
            "ID": case_id,
            "Año": case['year'],
            "Marca": case['manufacturer'],
            "Modelo": case['model'],
            "Precio": case['price'],
            "Similitud": round(sim_val, 4)
        })
    return rows

for qid in ["query_1", "query_2", "query_3"]:
    q = queries[qid]
    print(f"\n{'='*70}")
    print(f"Consulta {qid}: {q['year']} {q['manufacturer']} {q['model']} ({q['condition']}, {q['odometer']:,} km)")
    print(f"{'='*70}")
    
    rows_basic = run_retriever(casebase, q, retriever_basic, "Básico")
    rows_improved = run_retriever(casebase, q, retriever_improved, "Mejorado")
    
    comparison_df = pd.DataFrame(rows_basic + rows_improved)
    print(comparison_df.to_string(index=False))
```

El retriever mejorado con `fuel_sim` permite que los híbridos tengan una similitud parcial (0.7), lo que puede ampliar el pool de candidatos útiles.

- **query_3** (Honda Civic 2015, fair, 85000 km): Las mejoras son notables en el tratamiento de la condición. El retriever básico usa igualdad (fair = fair: 1.0, fair = good: 0.0), mientras que el mejorado asigna fair-good = 0.5, lo que permite recuperar coches en buen estado como alternativas razonables.

En general, las mejoras son más notables cuando la consulta tiene atributos que no coinciden exactamente con los casos de la base, ya que las métricas graduales capturan similitudes parciales que la igualdad exacta ignora.

## Parte 5: Adaptación de la solución (Reuse)

Una vez recuperados los casos más similares, la etapa de **Reuse** adapta su solución al nuevo problema. En nuestro dominio, la "solución" es el **precio** del vehículo.

Implementaremos una función de adaptación que **estima el precio** del vehículo consulta en base a los casos recuperados.

### 5.1 Adaptación simple: media ponderada de precios

```python
# Función de adaptación: copia directa del precio (sin modificación)

def adapt_price_weighted_mean(case: dict, query: dict) -> dict:
    """
    Adaptación trivial: devuelve el caso sin modificar el precio.
    """
    adapted = dict(case)
    return adapted

print("Función de adaptación básica definida")

# Para el reuser necesitamos una función de similitud (no un retriever)
# Usamos la misma attribute_value sim del retriever básico
base_sim = cbrkit.sim.attribute_value(
    attributes={
        "year":         year_sim_linear,
        "odometer":     odometer_sim_linear,
        "manufacturer": equality_sim,
        "fuel":         equality_sim,
        "transmission": equality_sim,
        "condition":    equality_sim,
        "model":        levenshtein_sim,
    },
    aggregator=cbrkit.sim.aggregator(pooling="mean", 
        pooling_weights={
        "year":         0.20,
        "odometer":     0.20,
        "manufacturer": 0.20,
        "fuel":         0.10,
        "transmission": 0.05,
        "condition":    0.15,
        "model":        0.10,
    })
)

reuser_basic = cbrkit.reuse.build(adapt_price_weighted_mean, base_sim)
print("Reuser básico construido")
```

```python
# Pipeline completo: primero retrieve, luego reuse
query = queries["query_1"]

# Paso 1: Retrieve
retrieval_result = cbrkit.retrieval.apply.apply_query(casebase, query, retriever_basic)

# Paso 2: Reuse (adaptación trivial) aplicado sobre el resultado del retrieval
reuse_result = cbrkit.reuse.apply_result(retrieval_result, reuser_basic)

step = reuse_result.steps[-1]
qstep = list(step.queries.values())[0]

print(f"{'ID':>4} {'Año':>5} {'Marca':>12} {'Modelo':>22} {'Precio':>8} {'Similitud':>10}")
print("-" * 70)
for case_id in qstep.ranking:
    sim_val = float(qstep.similarities[case_id].value)
    case = qstep.casebase[case_id]
    print(f"{case_id:>4} {case['year']:>5} {case['manufacturer']:>12} "
          f"{case['model']:>22} {case['price']:>8,} {sim_val:>10.4f}")
```

### **Ejercicio 5:** Implementa tu propia función de adaptación

La función de adaptación anterior solo considera año, kilometraje y condición. Mejórala añadiendo:

1. Una corrección cuando el **tipo de combustible** es diferente (ej., los coches eléctricos suelen ser más caros que los de gasolina equivalentes).
2. Una corrección cuando el **fabricante** es distinto (ej., marcas de lujo como BMW tienen precios más altos).
3. Prueba tu función de adaptación mejorada en las tres consultas y compara los precios estimados con la versión anterior.

```python
# Ejercicio 5: Función de adaptación mejorada

FUEL_PRICE_FACTOR = {
    "electric": 1.15,
    "hybrid":   1.05,
    "gas":      1.00,
    "diesel":   1.02,
    "other":    0.95,
}

MANUFACTURER_TIER = {
    "bmw":           1.20,
    "tesla":         1.25,
    "mercedes-benz": 1.20,
    "audi":          1.20,
    "lexus":         1.15,
    "subaru":        1.05,
    "volkswagen":    1.05,
    "honda":         1.00,
    "toyota":        1.00,
    "chevrolet":     0.90,
    "ford":          0.90,
    "nissan":        0.90,
    "hyundai":       0.88,
    "kia":           0.88,
}

def adapt_price_advanced_v2(case: dict, query: dict) -> dict:
    """
    Función de adaptación mejorada que ajusta el precio del caso recuperado
    en función de diferencias en año, kilometraje, condición, combustible y marca.
    """
    adapted = dict(case)
    base_price = case["price"]

    # Ajuste por año: depreciación de ~5% por año de diferencia
    year_diff = query["year"] - case["year"]
    year_factor = 1 + year_diff * 0.05

    # Ajuste por kilometraje: fracción del precio base por diferencia de km
    km_diff = case["odometer"] - query["odometer"]
    km_adjustment = km_diff * 0.002

    # Ajuste por condición
    condition_order = {"excellent": 4, "good": 3, "fair": 2, "salvage": 1}
    cond_diff = condition_order.get(query["condition"], 2) - condition_order.get(case["condition"], 2)
    condition_factor = 1 + cond_diff * 0.08

    # Ajuste por combustible
    query_fuel_factor = FUEL_PRICE_FACTOR.get(query["fuel"], 1.0)
    case_fuel_factor = FUEL_PRICE_FACTOR.get(case["fuel"], 1.0)
    fuel_factor = query_fuel_factor / case_fuel_factor

    # Ajuste por marca
    query_mfr_factor = MANUFACTURER_TIER.get(query["manufacturer"], 1.0)
    case_mfr_factor = MANUFACTURER_TIER.get(case["manufacturer"], 1.0)
    manufacturer_factor = query_mfr_factor / case_mfr_factor

    adapted_price = base_price * year_factor * condition_factor * fuel_factor * manufacturer_factor
    adapted_price += base_price * km_adjustment
    adapted["price"] = round(max(adapted_price, 0))

    return adapted

# Similitud mejorada para el reuser v2
improved_sim = cbrkit.sim.attribute_value(
    attributes={
        "year":         year_sim_exponential,
        "odometer":     odometer_sim_linear,
        "manufacturer": equality_sim,
        "fuel":         fuel_sim,
        "transmission": equality_sim,
        "condition":    condition_sim_table,
        "model":        model_sim_prefix,
    },
    aggregator=cbrkit.sim.aggregator(pooling="mean", 
        pooling_weights={
        "year":         0.20,
        "odometer":     0.20,
        "manufacturer": 0.20,
        "fuel":         0.10,
        "transmission": 0.05,
        "condition":    0.15,
        "model":        0.10,
    })
)

reuser_v2 = cbrkit.reuse.build(adapt_price_advanced_v2, improved_sim)

# Comparar ambas funciones de adaptación para las tres consultas
print(f"{'Consulta':<12} {'Adaptación básica':>20} {'Adaptación mejorada':>20}")
print("-" * 55)

for qid in ["query_1", "query_2", "query_3"]:
    q = queries[qid]

    # Retrieve primero
    ret_result = cbrkit.retrieval.apply.apply_query(casebase, q, retriever_improved)

    # Adaptación básica
    reuse_basic_r = cbrkit.reuse.apply_result(ret_result, reuser_basic)
    step_b = reuse_basic_r.steps[-1]
    qstep_b = list(step_b.queries.values())[0]
    top_id_b = list(qstep_b.ranking)[0]
    price_basic = qstep_b.casebase[top_id_b]["price"]

    # Adaptación mejorada
    reuse_v2_r = cbrkit.reuse.apply_result(ret_result, reuser_v2)
    step_v2 = reuse_v2_r.steps[-1]
    qstep_v2 = list(step_v2.queries.values())[0]
    top_id_v2 = list(qstep_v2.ranking)[0]
    price_v2 = qstep_v2.casebase[top_id_v2]["price"]

    print(f"{qid:<12} {price_basic:>20,} {price_v2:>20,}")
```

**Análisis del Ejercicio 5:**

La función de adaptación mejorada corrige el precio del caso recuperado considerando cinco dimensiones:

1. **Año**: Si la consulta es de un año más reciente que el caso, el precio sube (~5% por año), reflejando la depreciación temporal.
2. **Kilometraje**: Más kilómetros en el caso que en la consulta implica un ajuste al alza del precio estimado, ya que el coche consultado tiene menos desgaste.
3. **Condición**: Se ajusta proporcionalmente según la escala ordinal (excellent > good > fair > salvage).
4. **Combustible**: Los eléctricos e híbridos tienen un factor premium respecto a los de gasolina, reflejando la realidad del mercado.
5. **Marca**: Las marcas premium (BMW, Tesla, Audi) tienen un factor multiplicador mayor que las marcas económicas (Hyundai, Kia).

La adaptación básica simplemente copia el precio del caso más similar, lo cual es impreciso cuando hay diferencias significativas entre el caso recuperado y la consulta. La versión mejorada produce estimaciones más ajustadas al considerar estas correcciones multiplicativas.

## Parte 7: Preguntas de reflexión

Responde a las siguientes preguntas en celdas de texto adicionales:

**Q1.** ¿Qué ventajas tiene usar una tabla de similitud semántica para `condition` frente a la igualdad exacta? ¿Puedes pensar en algún dominio donde la igualdad exacta sea siempre la opción correcta?

**Q2.** Los pesos de los atributos modelan la importancia relativa de cada uno. Si el sistema está orientado a compradores que priorizan el ahorro de combustible, ¿cómo modificarías los pesos?

**Q3.** La función de adaptación asume una depreciación lineal por año y por kilómetro. ¿Es esta una suposición razonable? ¿Qué modelo de depreciación sería más realista?

**Q4.** ¿Cómo implementarías la fase de **Retain** en este sistema? ¿Bajo qué condiciones añadirías un nuevo caso a la base?

**Q5.** CBRkit soporta embedding semántico (`cbrkit.sim.embed`) para comparar textos. ¿En qué situaciones del dominio del coche podría ser útil? ¿Qué limitaciones tendría?

**Q1.** La tabla de similitud semántica para `condition` permite capturar que "excellent" y "good" son más parecidos entre sí que "excellent" y "salvage". Con igualdad exacta, cualquier diferencia (por mínima que sea) da similitud 0, perdiendo información valiosa sobre la proximidad entre categorías. Esto es especialmente útil cuando la base de casos es pequeña y no hay coincidencias exactas.

Un dominio donde la igualdad exacta sería siempre correcta sería el de los **grupos sanguíneos para transfusiones**: un paciente con grupo A+ no puede recibir sangre B+ (no hay "similitud parcial" que valga, la compatibilidad es binaria).

**Q2.** Si el sistema está orientado a compradores que priorizan el ahorro de combustible, se debería aumentar significativamente el peso del atributo `fuel` (por ejemplo, de 0.10 a 0.30) y también dar más peso a atributos correlacionados como el `year` (los coches más nuevos suelen ser más eficientes). Se podría reducir el peso de `model` y `manufacturer`, ya que al comprador le importa menos la marca concreta que la eficiencia. Los pesos podrían quedar: fuel=0.30, year=0.20, odometer=0.15, condition=0.15, manufacturer=0.10, model=0.05, transmission=0.05.

**Q3.** La depreciación lineal no es realista. En la práctica, los coches pierden más valor en los primeros años (un coche nuevo pierde ~20% al salir del concesionario y ~15% el primer año) y la depreciación se estabiliza con el tiempo. Un modelo más realista sería una **depreciación exponencial** del tipo `valor = precio_nuevo * e^(-λ * años)`, donde λ controla la velocidad de depreciación. Para el kilometraje ocurre algo similar: los primeros kilómetros deprecian más que los últimos (un coche con 200,000 km no vale mucho menos que uno con 180,000 km, pero la diferencia entre 10,000 y 30,000 km es significativa).

**Q4.** La fase de **Retain** añadiría nuevos casos a la base cuando aporten información que no está bien representada. Se podría implementar con las siguientes condiciones:

1. **Baja similitud máxima**: Si al buscar el caso más similar, la similitud es inferior a un umbral (por ejemplo, 0.6), significa que la base no tiene buenos representantes para esa consulta → se añade el caso.
2. **Error de predicción alto**: Si tras la fase de Reuse, el precio estimado difiere significativamente del precio real (cuando se conoce posteriormente), el caso debería incorporarse para mejorar futuras estimaciones.
3. **Nueva categoría**: Si aparece un fabricante, tipo de combustible o modelo no representado en la base.

También habría que implementar un mecanismo de **olvido** para eliminar casos obsoletos (coches muy antiguos que ya no son relevantes para el mercado actual).

**Q5.** El embedding semántico (`cbrkit.sim.embed`) podría ser útil para comparar el atributo `model`, ya que los nombres de modelos tienen significado semántico que las métricas de distancia de edición no capturan. Por ejemplo, "RAV4" y "CR-V" son ambos SUV compactos, pero tienen distancia de edición alta. Un embedding entrenado en texto automotriz podría captar que ambos pertenecen al mismo segmento.

**Limitaciones:**
- Los embeddings genéricos (tipo sentence-transformers) no están entrenados específicamente en nomenclatura automotriz, por lo que podrían no captar bien las relaciones entre modelos.
- Es computacionalmente más costoso que las métricas simples (Levenshtein, igualdad).
- Los nombres de modelos son muy cortos y a veces crípticos ("330i", "XLE"), lo que dificulta que los embeddings extraigan significado semántico útil.
- Requiere un modelo de embeddings como dependencia adicional, lo que puede ser excesivo para un dataset pequeño donde una tabla de similitud manual sería más precisa y controlable.
