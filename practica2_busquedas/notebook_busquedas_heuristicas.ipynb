#### Para el problema del puzle de 8  vamos a definir las siguientes funciones heurísticas:
* linear(node): cuenta el número de casillas mal colocadas respecto al estado final.
* manhattan(node): suma la distancia Manhattan desde cada casilla a la posición en la que debería estar en el estado final.
* max_heuristic(node): maximo de las dos anteriores
* sqrt_manhattan(node):  raíz cuadrada de la distancia Manhattan

```python
class Ocho_Puzzle(Problem):
    """Problema a del 8-puzzle.  Los estados serán tuplas de nueve elementos,
    permutaciones de los números del 0 al 8 (el 0 es el hueco). Representan la
    disposición de las fichas en el tablero, leídas por filas de arriba a
    abajo, y dentro de cada fila, de izquierda a derecha. 
    
    Las cuatro acciones del problema las representaremos mediante las cadenas:
    "Mover hueco arriba", 
    "Mover hueco abajo", 
    "Mover hueco izquierda" y
    "Mover hueco derecha"."""""

    def __init__(self, initial, goal=(1, 2, 3, 4, 5, 6, 7, 8, 0)):
        """ Define goal state and initialize a problem """
        self.goal = goal
        Problem.__init__(self, initial, goal)

    def actions(self, estado):
        pos_hueco = estado.index(0)
        accs = list()

        if pos_hueco >= 3:
            accs.append("Mover hueco arriba")
        if pos_hueco < 6:
            accs.append("Mover hueco abajo")
        if pos_hueco % 3 != 0:
            accs.append("Mover hueco izquierda")
        if pos_hueco % 3 != 2:
            accs.append("Mover hueco derecha")

        return accs

    def result(self, estado, accion):
        pos_hueco = estado.index(0)
        l = list(estado)
        def swap(x):
            l[pos_hueco], l[pos_hueco+x] = l[pos_hueco+x], l[pos_hueco]

        if accion in self.actions(estado):
            match accion:
                case "Mover hueco arriba":
                    swap(-3)
                case "Mover hueco abajo":
                    swap(3)
                case "Mover hueco izquierda":
                    swap(-1)
                case "Mover hueco derecha":
                    swap(1)

        return tuple(l)

    def linear(self, node):
        return sum(1 for s, g in zip(node.state, self.goal) if s != g and s != 0)

    def manhattan(self, node):
        total = 0
        for i, tile in enumerate(node.state):
            if tile != 0:
                goal_pos = self.goal.index(tile)
                total += abs(i // 3 - goal_pos // 3) + abs(i % 3 - goal_pos % 3)
        return total

    def sqrt_manhattan(self, node):
        from math import sqrt
        total = 0
        for i, tile in enumerate(node.state):
            if tile != 0:
                goal_pos = self.goal.index(tile)
                total += abs(i // 3 - goal_pos // 3) + abs(i % 3 - goal_pos % 3)
        return sqrt(total)

    def max_heuristic(self, node):
        return max(self.linear(node), self.manhattan(node))

    def h(self, node):
        """ Return the heuristic value for a given state."""
        return 1
    
    def coste_de_aplicar_accion(self, s, a):
        return 1
```

Orden de mayor a menor informada: max_heuristic = manhattan > linear > sqrt_manhattan

max_heuristic: es la más informada (aunque igualada con manhattan) porque devuelve el máximo de manhattan y linear. Como manhattan > linear; maxheuristic = manhattan.

manhattan: estima cuántos movimientos mínimos necesita cada pieza para llegar a su posición, lo que da valores más altos y cercanos al coste real que simplemente contar piezas mal colocadas.

linear: solo cuenta cuántas piezas están fuera de sitio, ignorando cuánto les falta moverse. Da valores más bajos que manhattan y por tanto guía menos la búsqueda.

sqrt_manhattan: es la menos informada porque aplicar la raíz cuadrada reduce artificialmente los valores de la heurística, alejándolos del coste real y haciendo que A* explore más nodos.
